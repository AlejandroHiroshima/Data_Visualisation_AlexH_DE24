```python
import pandas as pd
import duckdb

df = pd.read_excel("data/resultat-ansokningsomgang-2024.xlsx", sheet_name = "Tabell 3", skiprows=5)
df.head()
```

```python
decisions = df["Beslut"].value_counts()
approved, total = decisions["Beviljad"], decisions.sum()
approved, total
```

```python
df_regions = duckdb.query(
    """
    SELECT län, CAST(COUNT_IF(beslut = 'Beviljad') AS INTEGER) AS Beviljade
    FROM df
    WHERE län != 'Flera kommuner'
    GROUP BY
    län
    ORDER BY beviljade DESC
    """
).df()
df_regions
```
